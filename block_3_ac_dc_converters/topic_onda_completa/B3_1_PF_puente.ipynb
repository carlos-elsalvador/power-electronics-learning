# Problema 1 — Factor de potencia de un rectificador monofásico

![Figura 1. Rectificador monofásico de onda completa con filtrado por condensador de la tensión de salida.](figures/Fig1_puente.png)

Un rectificador monofásico no controlado de onda completa, con filtro capacitivo, alimenta una red de distribución DC. Sus datos son:

- Tensión eficaz de entrada: $V_{RMS}=230\ \mathrm{V}$.
- Frecuencia: $f=50\ \mathrm{Hz}$.
- Potencia activa: $P=2\ \mathrm{kW}$.
- Factor de desplazamiento: $\cos\varphi_1=1$.

El espectro de la corriente de entrada, normalizado respecto
de la amplitud de la componente fundamental, se muestra en
la siguiente tabla.


| Orden $n$ | 3 | 5 | 7 | 9 | 11 | 13 | 15 | 17 |
|---|---:|---:|---:|---:|---:|---:|---:|---:|
| $\frac{I_{n}}{I_{1}}$ (%) | 81,2 | 29,8 | 9,4 | 6,1 | 4,4 | 2,8 | 0,79 | 0,39 |


Se considera que la corriente **no es sinusoidal**, que no
contiene componente continua ($I_0=0$) y que los armónicos
no incluidos en la tabla son despreciables. La potencia indicada corresponde a la potencia activa de entrada. Además, se considera una tensión de entrada
sinusoidal ideal: $v_s(t)=V_s\sin(\omega t)$, donde $V_s=\sqrt{2}V_{RMS}$.

# **Determinar el factor de potencia del sistema.**


## Paso 1. Definición del factor de potencia

El factor de potencia es el cociente entre la potencia activa y la potencia aparente:

$$PF=\frac{P}{S}=\frac{P}{V_{RMS} I_{RMS}}$$

Donde $I_{RMS}$ es el valor eficaz **total** de la corriente de entrada, no solamente el de su componente fundamental (**recordar: la corriente no es sinusoidal**.) Por otra parte, la tensión de entrada es $v_s(t)=\sqrt{2}V_{RMS}\sin(\omega t) =\sqrt{2} \cdot 230\sin(\omega t)$, con frecuencia angular $\omega=2\pi f=100\pi\ \mathrm{rad/s}.$  El condensador del rectificador favorece una corriente de entrada pulsante, que contiene componentes armónicas.

## Paso 2. Potencia activa, factor de desplazamiento y Factor de distorsión

Utilizando los valores $I_{n}$ representamos la corriente mediante una serie de Fourier:

$$i_s(t)=\sum_{n=1}^{\infty}I_{n}\sin(n\omega t-\varphi_n).$$

 Por ortogonalidad de las funciones sinusoidales, al calcular

$$P=\frac{1}{T}\int_0^T v_s(t)i_s(t)\,dt,$$

los productos entre la tensión fundamental y las corrientes de frecuencia distinta tienen valor medio nulo. Por tanto:

$$P=V_{RMS} I_{1, RMS}\cos\varphi_1$$

Sustituyendo en la definición del factor de potencia:

$$PF=\frac{I_{1,RMS}}{I_{RMS}}\cos\varphi_1.$$

El cociente $\frac{I_{1,RMS}}{I_{RMS}}$ y el factor $\cos\varphi_1,$ reciben el nombre de **factor de distorsión** y **factor de desplazamiento**, respectivamente.

**Nota**: Del enunciado del problema, se sabe que el factor de desplazamiento es igual  a $\cos\varphi_1=1.$ Esto significa que la componente fundamental de la corriente está en fase con la tensión de entrada. Sin embargo, esto no implica que la corriente total sea sinusoidal ni que el factor de potencia sea unitario.

## Paso 3. Cálculo de la distorsión armónica total (THD)

La distorsión armónica total de corriente se define como:

$$THD_I=\frac{\sqrt{\sum_{n=2}^{\infty} I_{n}^2}}{I_{1}}=\sqrt{\sum_{n=2}^{\infty}\left(\frac{I_{n, RMS}}{I_{1,RMS}}\right)^2}.$$

Observar que la tabla proporciona relaciones de amplitudes, estas coinciden con las relaciones de valores eficaces, porque el factor $\sqrt{2}$ se cancela. Se convierten los porcentajes a valores decimales y se suman sus cuadrados, $THD_I^2=0{,}7635.$ Por tanto, $THD_I=\sqrt{0{,}7635}=0{,}8738,$ y, finalmente, como porcentaje $THD_I=87,38\%.$ El valor eficaz conjunto de los armónicos equivale al $87{,}38 \%$ del valor eficaz fundamental. De nuevo: **no se deben sumar directamente los porcentajes**.

## Paso 4. Relación entre THD y factor de potencia

Aprovecharemos este ejemplo para demostrar la relación que existe entre factor potencia ($PF$) y distorsión armónica ($THD$). Por el camino nos volvemos a encontrar de nuevo con  el factor de distorsión ($FDis$) y el factor de desplazamiento ($FD$).

Por ortogonalidad (teorema de Parseval), el cuadrado del valor eficaz total es la suma de los cuadrados de los valores eficaces individuales:

$$I_{RMS}^2=I_{1,RMS}^2+\sum_{n=2}^{\infty}I_{n, RMS}^2.$$

Dividiendo por la fundamental $I_{1, RMS}^2$:

$$\left(\frac{I_{RMS}}{I_{1, RMS}}\right)^2=1+THD_I^2.$$

Por tanto, el factor de distorsión (FDis) es:

$$FDis=\frac{I_{1,RMS}}{I_{RMS}}=\frac{1}{\sqrt{1+THD_I^2}}.$$

Ahora podemos reescribir el factor de potencia en función de THD:

$$PF=FD\cdot FDis=\frac{\cos\varphi_1}{\sqrt{1+THD_I^2}}.$$

El factor de distorsión indica qué fracción del valor eficaz total corresponde a la componente fundamental, $ FDis=\frac{1}{\sqrt{1+0,7635}}=0,753.$ Como el factor de desplazamiento es unitario, $\cos\varphi_1 = 1$, **el factor de potencia es aproximadamente 0,753**.

## Paso 5. Comprobación mediante potencia y corriente eficaz

La potencia activa permite determinar la corriente fundamental:

$$I_{1, RMS}=\frac{P}{V_{RMS}\cos\varphi_1}=\frac{2000}{230}=8,7\ \mathrm{A}.$$

La corriente eficaz total resulta:

$$I_{RMS}=I_{1, RMS}\sqrt{1+THD_I^2}\approx 11,55\ \mathrm{A}.$$

La potencia aparente es:

$$S=V_{RMS} I_{RMS}\approx 2656\ \mathrm{VA}.$$

Finalmente:

$$PF=\frac{2000}{2656}\approx\boxed{0,753}.$$

Los dos métodos proporcionan el mismo resultado. Los datos de potencia y tensión no eran necesarios para el primer método, pero permiten calcular las corrientes absolutas y comprobar el resultado.
