# Ejercicio 1. Análisis de Serie de Fourier de onda cuadrada bipolar con ángulo eléctrico como variable independiente

**Objetivo:** familiarizarse con la representación de la tensión de salida de un inversor mediante series de Fourier. Consideremos un inversor monofásico en puente completo, alimentado por una fuente de tensión continua $V_{DC}$. Los interruptores se controlan para generar una forma de onda que corresponde a una señal cuadrada bipolar:

$$
v_x(\theta)=
\begin{cases}
+V_{DC}, & 0<\theta<\pi\\
-V_{DC}, & \pi<\theta<2\pi
\end{cases}
$$

Donde: $\theta=\omega_0t$. Se pide:

**a)** Representar gráficamente la señal durante un período.

**b)** Determinar si la señal presenta simetría impar y simetría de media onda.

**c)** Calcular el valor promedio:

$$
V_{avg}=\frac{1}{2\pi}
\int_0^{2\pi}v_x(\theta)\,d\theta
$$

**d)** Calcular los coeficientes $a_n$ y $b_n$ de Fourier, utilizando:

$$
a_n=\frac{1}{\pi}
\int_0^{2\pi}v_x(\theta)\cos(n\theta)\,d\theta
$$

$$
b_n=\frac{1}{\pi}
\int_0^{2\pi}v_x(\theta)\sin(n\theta)\,d\theta
$$

**e)** Obtener la expresión general de la serie de Fourier.

**f)** Identificar los primeros cinco armónicos presentes en la señal.

**g)** Explicar por qué una onda cuadrada no contiene armónicos pares.

# Ejercicio 2. Modificación de la onda cuadrada mediante un ángulo de conmutación

**Objetivo:** analizar cómo afecta el ángulo de conmutación al contenido armónico de la señal. Consideremos nuevamente la forma de onda del ejercicio anterior. Ahora modificamos la estrategia de conmutación introduciendo intervalos de tensión cero al principio y al final de cada semiciclo. La nueva señal queda definida por:

$$
v_x(\theta)=
\begin{cases}
+V_{DC}, & \delta<\theta<\pi-\delta\\
-V_{DC}, & \pi+\delta<\theta<2\pi-\delta\\
0, & \text{resto del período}
\end{cases}
$$

Donde: $0<\delta<\frac{\pi}{2}$. Se pide:

**a)** Representar gráficamente la nueva forma de onda para un período completo.

**b)** Comprobar que la señal presenta simetría de media onda:

$$
v_x(\theta+\pi)=-v_x(\theta)
$$

**c)** Determinar si presenta simetría impar.

**d)** Calcular el valor promedio de la señal.

**e)** Determinar cuáles de los coeficientes $a_n$ y $b_n$ son necesariamente iguales a cero.

**f)** Calcular el coeficiente $b_n$ mediante:

$$
b_n=\frac{1}{\pi}
\int_0^{2\pi}v_x(\theta)\sin(n\theta)\,d\theta
$$

**g)** Demostrar que el resultado para los armónicos impares es:

$$
\boxed{
b_n=\frac{4V_{DC}}{n\pi}\cos(n\delta)
}
$$

Y que para los armónicos pares:

$$
\boxed{b_n=0}
$$

**h)** Escribir los primeros cinco términos no nulos de la serie de Fourier.

# Ejercicio 3. Eliminación del tercer armónico

**Objetivo:** utilizar el análisis de Fourier para determinar el ángulo de conmutación que elimina el tercer armónico. Utilizando la expresión obtenida en el ejercicio anterior:$b_n=\frac{4V_{DC}}{n\pi}\cos(n\delta)$. Se pide:

**a)** Escribir la expresión del coeficiente correspondiente a la componente fundamental, $b_1$

**b)** Escribir la expresión del coeficiente correspondiente al tercer armónico, $b_3$

**c)** Imponer la condición necesaria para eliminar el tercer armónico,$b_3=0
$

**d)** Resolver la ecuación y determinar el valor de $\delta$ en grados eléctricos y radianes.

**f)** Calcular el coeficiente correspondiente al quinto armónico.

**g)** Explicar por qué eliminar el tercer armónico no significa que la señal se haya convertido en una sinusoide.

# Ejercicio 4. Influencia del ángulo de conmutación sobre los armónicos

**Objetivo:** comprender que el ángulo de conmutación modifica simultáneamente la amplitud de diferentes armónicos.

Consideremos el mismo inversor y la siguiente expresión:

$$
b_n=\frac{4V_{DC}}{n\pi}\cos(n\delta)
$$

Se propone estudiar tres condiciones de funcionamiento:

| Caso | Ángulo de conmutación |
|:---:|:---:|
| A | $0^\circ$ |
| B | $15^\circ$ |
| C | $30^\circ$ |

Se pide:

**a)** Calcular los coeficientes $b_1$, $b_3$, $b_5$ y $b_7$ para cada uno de los tres casos.

**b)** Completar una tabla con los resultados normalizados respecto a $V_{DC}$:

$$
\frac{b_n}{V_{DC}}
$$

**c)** Representar gráficamente la evolución de los coeficientes de los diferentes armónicos en función del ángulo de conmutación.

**d)** Identificar qué sucede con la componente fundamental cuando aumenta el ángulo de conmutación.

**e)** Explicar por qué el ángulo de $30^\circ$ resulta especialmente interesante desde el punto de vista de la eliminación de armónicos.

**f)** Comparar las tres formas de onda y discutir cuál presenta mejores características para obtener una tensión sinusoidal mediante un filtro paso bajo.

# Ejercicio 5. Aplicación: diseño de un inversor para eliminar el tercer armónico

**Objetivo:** resolver un problema de diseño aplicando los conceptos matemáticos estudiados.

Se dispone de un inversor monofásico en puente completo cuya tensión de alimentación es:

$$
V_{DC}=200\text{ V}
$$

La frecuencia fundamental de salida es:

$$
f_0=50\text{ Hz}
$$

Se desea generar una tensión alterna utilizando la estrategia de conmutación con un único ángulo $\delta$.

El objetivo del diseño es eliminar completamente el tercer armónico.

Se pide:

**a)** Calcular la frecuencia angular fundamental:

$$
\omega_0=2\pi f_0
$$

**b)** Determinar el período de la señal:

$$
T=\frac{1}{f_0}
$$

**c)** Calcular el ángulo de conmutación necesario para eliminar el tercer armónico.

**d)** Determinar los instantes de conmutación correspondientes a:

$$
\delta,\quad
\pi-\delta,\quad
\pi+\delta,\quad
2\pi-\delta
$$

Expresar los resultados tanto en grados eléctricos como en milisegundos.

**e)** Calcular la amplitud de la componente fundamental:

$$
V_1=|b_1|
$$

**f)** Calcular las amplitudes de los armónicos quinto y séptimo:

$$
V_5=|b_5|
$$

$$
V_7=|b_7|
$$

**g)** Determinar la relación entre la amplitud del quinto armónico y la fundamental:

$$
\frac{|b_5|}{|b_1|}
$$

Expresar el resultado en porcentaje.

**h)** Explicar qué ventajas presenta esta estrategia de conmutación respecto a una onda cuadrada convencional.
