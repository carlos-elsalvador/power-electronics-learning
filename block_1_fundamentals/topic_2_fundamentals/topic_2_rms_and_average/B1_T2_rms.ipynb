## 0. Nivel DC

El tipo de forma de onda que se analizará en electrónica de potencia será el de régimen permanente, (PSS, *Periodic Stationary State*). Serán señales periódicas que pueden descomponerse en una parte continua (Nivel DC) y en otra alterna (Nivel AC). El nivel DC es el promedio (o valor medio) de la señal calculado a lo largo de un periodo. Matemáticamente, sea $v(t)$ una señal periódica de tensión, con periodo $T$, el valor medio se obtiene mediante la expresión:
$$
\left\langle v(t) \right\rangle
\quad =
\frac{1}{T}
\int_0^T v(t)\,dt
$$
Como se verá más adelante, una señal periódica se puede descomponer en serie de Fourier, como suma de señales sinusoidales, tipo seno o coseno, más un término constante. El término constante es el nivel DC y se corresponde con el primer término del desarrollo en serie de Fourier.

Como la integral corresponde al área bajo la curva, también el valor promedio se puede calcular como  el cociente entre él área y el periodo
$$
\left\langle v(t) \right\rangle
\quad =
\frac{Área}{T}
$$

## 1. Potencia instantánea

La potencia instantánea asociada a una tensión $v(t)$ y una corriente $i(t)$ se define como $p(t) = v(t)i(t).$ Para una resistencia $R$ y, utilizando la ley de Ohm $v(t)=Ri(t),$ la potencia consumida en cada instante se escribe como:

$$
p(t)=Ri^2(t)
$$

También podemos escribir

$$
p(t)=\frac{v^2(t)}{R}
$$


* $\quad$ Si $p(t)>0$, el elemento está absorbiendo o **disipando potencia**.

* $\quad$Si $p(t)<0$, el elemento está suministrando potencia al circuito.

## 2. Potencia media disipada

Para señales DC, la potencia disipada no depende del tiempo. Sin embargo, para señales AC la potencia sí depende del tiempo. Para señales periódicas, el parámetro que caracteriza la **potencia disipada** a efectos de consumo y calentamiento de los dispositivos a lo largo del tiempo es la **potencia media disipada** en un periodo:

$$
\left\langle p(t) \right\rangle
\quad =
\frac{1}{T}
\int_0^T p(t)\,dt
$$


## 3. Valor eficaz o rms (root mean square)

Conceptualmente, el valor eficaz de una seña AC de tensión o una corriente se define como el valor que habría que aplicar en continua para que la potencia media disipada fuera la misma.

* Para una tensión variable $v(t)$:

$$
\left\langle p(t) \right\rangle
\quad =
\frac{1}{T}
\int_0^T v(t)i(t)\,dt
\quad =
\frac{1}{T}
\int_0^T \frac{v^2(t)}{R}\,dt
\quad =
\frac{1}{R}·\frac{1}{T}
\int_0^T v^2(t)\,dt
\quad = \frac{V^2_{\mathrm{RMS}}}{R}
$$

* Para una corriente variable $i(t)$:

$$
\left\langle p(t) \right\rangle
\quad =
\frac{1}{T}
\int_0^T v(t)i(t)\,dt
\quad =
\frac{1}{T}
\int_0^T Ri^2(t)\,dt
\quad =
R·\frac{1}{T}
\int_0^T i^2(t)\,dt
\quad = R·I^2_{\mathrm{RMS}}
$$



Por tanto,

$$
\boxed{
I_{\mathrm{RMS}}
=
\sqrt{
\frac{1}{T}
\int_0^T i^2(t)\,dt
}
}
\qquad
\boxed{
V_{\mathrm{RMS}}
=
\sqrt{
\frac{1}{T}
\int_0^T v^2(t)\,dt
}
}
$$

## Ejemplo 1 — Señal continua

Consideremos una corriente constante:

$$
i(t)=2\ \mathrm{A}
$$

Como la corriente no cambia con el tiempo:

$$
I_{\mathrm{RMS}}
=
\sqrt{
\frac{1}{T}
\int_0^T 2^2\,dt
}
\quad = \quad
\sqrt{
\frac{1}{T}
4T
}
$$


Finalmente $I_{\mathrm{RMS}}=2\ \mathrm{A}$. Para una resistencia de $R=5\,\Omega$:

$$
\left\langle p(t) \right\rangle
=
R· I_{\mathrm{RMS}}^2
=
5(2)^2
=
20\ \mathrm{W}
$$


## Ejemplo 2 — Señal sinusoidal


![Figura 1. Corriente Alterna](figures/1_ca.jpeg)

Consideremos una tensión sinusoidal $v(t)=V_{\mathrm{p}}\sin(\omega t)$ donde $V_{\mathrm{p}}$ es el valor máximo o amplitud y $\omega$ esl la frecuencia angular con la que se repite la señal. El valor eficaz de la tensión es

$$
V_{\mathrm{RMS}}
=
\sqrt{
\frac{1}{T}
\int_0^T
v^2(t)\,dt
}
\quad =\quad
\sqrt{
\frac{1}{T}
\int_0^T
V_{\mathrm{p}}^2\sin^2(\omega t)\,dt
}
\quad =\quad
\sqrt{
V_{\mathrm{p}}^2\left\langle\sin^2(\omega t)\right\rangle
}
$$

El valor medio de $\sin^2(\omega t)$ durante un periodo es

$$
\left\langle\sin^2(\omega t)\right\rangle
\quad =\quad
\frac{1}{T} \int_0^T
\sin^2(\omega t)\,dt
\quad =\quad
\frac{1}{T} \int_0^T
\frac{1}{2}(1-\cos(2\omega t))\,dt
=
\frac{1}{2}
$$

Por tanto:

$$
\boxed{
V_{\mathrm{RMS}}
=
\frac{V_{\mathrm{p}}}{\sqrt{2}}
}
$$

El valor eficaz de una señal de alterna sinusoidal es la amplitud dividida por $\sqrt{2}$.

```python
# Parámetros de la señal
Vp          = 325       # V
f           = 50        # Hz
T           = 1 / f     # S
```

```python
# Parámetros de la simulación
NT          = 2         # Número de periodos
N           = 2000      # Total de muestras
```

```python
# Generación de tiempo y tensión
t           = np.linspace(0, NT*T, N, endpoint=False)
v_seno        = Vp * np.sin(2*np.pi*f*t)
```

```python
# Tensión rms
Vrms        = np.sqrt(np.mean(v_seno**2))
```

```python
# Imprimir resultados
print(f"V_p = {Vp:.2f} V")
print(f"V_RMS numérico = {Vrms:.4f} V")
print(f"V_RMS teórico  = {Vp/np.sqrt(2):.4f} V")
```

```python
# PLOT
plt.plot(t*1e3, v_seno)
plt.xlabel("Tiempo [ms]")
plt.ylabel("Tensión [V]")
plt.title("Señal sinusoidal")
plt.show()
```

## Ejemplo 3 — Señal con componente DC y AC

![Figura 2. Potencia](figures/2_ca.png)

Consideremos la señal de tensión $v(t)=V_{\mathrm{DC}}+V_{\mathrm{p}}\sin(\omega t)$.

El valor eficaz al cuadrado es:

$$
V_{\mathrm{RMS}}^2
\quad =\quad
\left\langle v^2(t)\right\rangle
\quad =\quad
\left\langle
\left(
V_{\mathrm{DC}}+
V_{\mathrm{p}}\sin(\omega t)
\right)^2
\right\rangle
$$

Por tanto:

$$
V_{\mathrm{RMS}}^2
=
V_{\mathrm{DC}}^2
+
2V_{\mathrm{DC}}V_{\mathrm{p}}
\left\langle\sin(\omega t)\right\rangle
+
V_{\mathrm{p}}^2
\left\langle\sin^2(\omega t)\right\rangle
$$

Donde el valor medio del término sinusoidal se anula, $\left\langle\sin(\omega t)\right\rangle=0$, y el del término cuadrático es una constante,  $\left\langle\sin^2(\omega t)\right\rangle=\frac{1}{2}$. Sustituyendo, obtenemos:

$$
\boxed{
V_{\mathrm{RMS}}^2
=
V_{\mathrm{DC}}^2+
\frac{V_{\mathrm{p}}^2}{2}
}
$$

Este resultado es general. El valor eficaz al cuadrado de una señal que combina un nivel de continua con un nivel de alterna, siempre es la suma de la señal de continua al cuadrado más el valor eficaz al cuadrado de la componente de alterna.

##  Ejemplo 4.1 —Valor medio de una onda de tensión cuadrada bipolar

Consideremos una señal periódica que toma solamente dos valores. La señal se repite cada periodo $T$ durante tiempos iguales:

$$
v(t)=
\begin{cases}
+V, & 0 \leq t < T/2 \\
-V, & T/2 \leq t < T
\end{cases}
$$


* $+V$: Durante la primera mitad del periodo tenemos una tensión positiva, el área correspondiente es: $A_+ = V·\frac{T}{2}$

* $-V$: Durante la segunda mitad tenemos una tensión negativa, su área es: $A_- = -V·\frac{T}{2}.$

* Las dos áreas se cancelan: $A_{\mathrm{total}}= A_+ + A_- = \quad
V·\frac{T}{2} - V·\frac{T}{2} = \quad 0 $

Por tanto, la tensión promedio $ \left\langle v(t) \right\rangle = 0$, la onda cuadrada bipolar tiene valor medio cero. Podría pensarse que la señal no produce potencia, pero esto no es cierto. Si se conecta la señal a una resistencia, tanto $+V$ como $-V$ producen calentamiento.

La potencia en una resistencia es $p(t)=\frac{v^2(t)}{R}$. Se observa que al elevar la tensión al cuadrado para la primera mitad, $(+V)^2=V^2$ y para la segunda mitad, $(-V)^2=V^2$, el signo desaparece. Por eso, para estudiar la potencia se utiliza el valor eficaz y no el valor medio.

## Ejemplo 4.2 — Valor eficaz de una onda de tensión cuadrada bipolar
Consideremos una señal periódica que toma solamente dos valores. La señal se repite cada periodo $T$ durante tiempos iguales:

$$
v(t)=
\begin{cases}
+V, & 0 \leq t < T/2 \\
-V, & T/2 \leq t < T
\end{cases}
$$

El valor eficaz es

$$
V_{\mathrm{RMS}}
=
\sqrt{
\frac{1}{T}
\int_0^T
v^2(t)\,dt
}
\quad = \quad
\sqrt{\left\langle v^2(t)\right\rangle
}
\quad = \quad
\sqrt{\frac{\text{área durante un periodo}}{\text{periodo}}}
$$

Elevando la señal al cuadrado, para la parte positiva y para la parte negativa, $(+V)^2=V^2$ y $(-V)^2=V^2$, respectivamente. La señal $v^2(t)=V^2$ es constante durante todo el periodo $T$. Por tanto, el área bajo la señal es simplemente el área de un rectángulo, $ \text{Área}=\text{base}\times\text{altura} = T V^2$. Para una señal periódica:

$$
\left\langle v^2(t)\right\rangle
\quad = \quad
\frac{TV^2}{T}
\quad = \quad
V^2
$$

entonces:

$$
V_{\mathrm{RMS}}
=
\sqrt{V^2}
$$

Por tanto:

$$
\boxed{
V_{\mathrm{RMS}}=V
}
$$

Es decir, para una onda cuadrada bipolar simétrica, el valor eficaz
es igual a su amplitud.


```python
# Onda cuadrada bipolar. La expresión np.where(condición, valor_si_verdadero, valor_si_falso) genera un array numpy.
# Aquí la condición es: (t%T) < (D·T).Es decir:
#  *   Se toma cada instante de tiempo t, array de longitud length(t).
#  *   Se calcula en qué parte del periodo está (t % T).
#  *   Si está dentro del intervalo (< T/2), se asigna +V.
#  *   Si está fuera, se asigna –V.
# Resultado: una onda rectangular bipolar de amplitud ±V.
Vp          = 10
v           = np.where((t % T) < T/2, Vp, -Vp)
```

```python
# Tensión rms
Vrms        = np.sqrt(np.mean(v**2))
```

```python
# Imprimir resultados
print(f"V_p = {Vp:.2f} V")
print(f"V_RMS numérico = {Vrms:.4f} V")
print(f"V_RMS teórico  = {Vp:.4f} V")
```

```python
# PLOT
plt.plot(t*1000, v)
plt.xlabel("Tiempo [ms]")
plt.ylabel("Tensión [V]")
plt.title("Señal onda cuadrada bipolar")
plt.show()
```

##  Ejemplo 4.3 —Valor medio de una onda cuadrada unipolar con ciclo de trabajo

Consideremos una señal periódica, con periodo $T$. Durante un tiempo de duración $DT$, ($0 < D < 1$) la señal presenta un nivel te tensión alto, $+V$ y el resto del tiempo, $(1-D)T$, vale cero.

$$
v(t)=
\begin{cases}
+V, & 0 \leq t < DT \\
0, & (1-D) \leq t < T
\end{cases}
$$


* $+V$: Durante la primera parte del periodo tenemos una tensión positiva, el área correspondiente es: $A_+ = V·DT$

* $0$: Durante la segunda parte tenemos tensión cero, su área es: $A_- = 0.$

* Las dos áreas se suman: $A_{\mathrm{total}}= A_+ + A_- = \quad
V·DT + 0 = \quad 0 $

Por tanto, la tensión promedio $ \left\langle v(t) \right\rangle = DV$, la **onda cuadrada unipolar con ciclo de trabajo D** tiene valor medio $DV$.

```python
# Onda cuadrada unipolar. La expresión np.where(condición, valor_si_verdadero, valor_si_falso) genera un array numpy.
# Aquí la condición es: (t%T) < (D·T).Es decir:
#  *   Se toma cada instante de tiempo t, array de longitud length(t).
#  *   Se calcula en qué parte del periodo está (t % DT).
#  *   Si está dentro del intervalo (< DT), se asigna +V.
#  *   Si está fuera, se asigna 0.
# Resultado: una onda rectangular unibipolar de amplitud V y ciclo de trabajo D.
D             = 0.25
Vp            = 10
v             = np.where((t % T) < D*T, Vp, 0)
```

```python
# Tensión media
Vmean        = np.mean(v)
```

```python
# Imprimir resultados
print(f"D = {D:.2f}")
print(f"V_p = {Vp:.2f} V")
print(f"V_avg numérico = {Vmean:.4f} V")
print(f"V_avg teórico  = {D*Vp:.4f} V")
```

```python
# PLOT
plt.plot(t*1000, v)
plt.xlabel("Tiempo [ms]")
plt.ylabel("Tensión [V]")
plt.title("Señal onda cuadrada unipolar con ciclo de trabajo D")
plt.show()
```

## Ejemplo 5.1 — Valor eficaz de una señal AC triangular

Considere una señal de corriente alterna triangular, $i(t),$ cuya amplitud oscila entre  $-i_{AC}$ e $+i_{AC}.$ Esta corriente es una señal de alterna pura, ya que su valor medio es cero, $I_{\mathrm{DC}}=0$. Para una amplitud $i_{AC} = 12A,$ la amplitud pico-pico es de $24A.$ Por simetría, para calcular el valor eficaz, podemos hacer el promedio para medio periodo, ya que el área de la señal al cuadrado durante los dos semiperiodos es la misma


$$
I_{\mathrm{RMS}}
\quad =
\frac{1}{T}
\int_0^T
i^2(t)\,dt
\quad =
\frac{2}{T}
\int_0^{T/2}
i^2(t)\,dt
$$

Para evaluar la integral en un semiperíodo, escribimos explícitamente la corriente en función del tiempo. Es la ecuación de una recta, teniendo en cuenta que $i(0) = -i_{AC},  i\left(\frac{T}{2}\right) = i_{AC}.$ Así, para $0 \le t \le T/2$:



$$
i(t) = -i_{AC} + \frac{4 i_{AC}}{T}\, t
$$



En general, para una señal lineal que cambia desde $-i_{AC}$ en $t = 0$ hasta $i_{AC}$ en $t = T'$:


$$
i(t) = -i_{AC} + \frac{2 i_{AC}}{T'}\, t
$$



Se utiliza esta notación para obtener un resultado más general en este ejemplo, que pueda aprovecharse en un futuro, y no restringir al caso particular de que
$T' = T/2$. Así el valor cuadrático de la corriente es:
$$
i^2(t) = i_{AC}^2 + \frac{4 i_{AC}^2}{T'^2}\, t^2 - \frac{4 i_{AC}^2}{T'}\, t
$$



El valor eficaz se obtiene como:



$$
I_{RMS}^2 = \frac{2}{T} \int_0^{T/2} i^2(t)\, dt
           = \frac{1}{T'} \int_0^{T'}
           \left(
               i_{AC}^2
               + \frac{4 i_{AC}^2}{T'^2} t^2
               - \frac{4 i_{AC}^2}{T'} t
           \right) dt
$$

Evaluando la integral:



$$
I_{RMS}^2 =
\frac{1}{T'}
\left[
    i_{AC}^2\, t
    + \frac{4 i_{AC}^2}{T'^2} \frac{t^3}{3}
    - \frac{4 i_{AC}^2}{T'} \frac{t^2}{2}
\right]_{0}^{T'}
\quad = \quad
\frac{1}{T'}
\left(
    i_{AC}^2 T'
    + \frac{4 i_{AC}^2 T'^3}{3 T'^2}
    - \frac{2 i_{AC}^2 T'^2}{T'}
\right)
\quad = \quad
\frac{i_{AC}^2}{3}
$$






Para esta señal, el cálculo del valor eficaz conduce a:


$$
\boxed{
I_{\mathrm{RMS}}
=
\frac{i_{AC}}{\sqrt{3}}
}
$$

```python
# Amplitud de la señal de corriente
i_AC    = 12
```

```python
tau     = t % T
i       = np.where(tau < T/2, -i_AC + (4*i_AC/T)*tau, +i_AC - (4*i_AC/T)*(tau - T/2))
```

```python
# Corriente rms
Irms        = np.sqrt(np.mean(i**2))
```

```python
# Imprimir resultados
print(f"i_AC = {i_AC:.2f} A")
print(f"I_RMS numérico = {Irms:.4f} A")
print(f"I_RMS teórico  = {i_AC/np.sqrt(3):.4f} A")
```

```python
# PLOT
plt.plot(t*1000, i)
plt.xlabel("Tiempo [ms]")
plt.ylabel("Corriente [A]")
plt.title("Señal de corriente triangular")
plt.show()
```

##  Ejemplo 5.2 — Valor medio de una corriente triangular general

Considere una señal de corriente triangular, $i(t)$, periodica, $T$, donde  la corriente oscila entre un valor mínimo $i_{\min}$ y un valor máximo $i_{\max}$, con un ripple $ \Delta i = i_{\max} - i_{\min}.$ Durante la primera parte del tiempo, de duración $T_{\mathrm{on}}$, la corriente crece linealmente desde $i_{\min}$ hasta $i_{\max}$. Durante el resto del tiempo, de duración $T_{\mathrm{off}}$, la corriente decrece linealmente. La señal se repite cada periodo $T = T_{\mathrm{on}} + T_{\mathrm{off}}$

$$
i(t)=
\begin{cases}
i_{\min} + \frac{\Delta i}{T_{\mathrm{on}}}\, t, \qquad & 0 \le t \le T_{\mathrm{on}}\\
i_{\max} - \frac{\Delta i}{T_{\mathrm{off}}}\, (t - T_{\mathrm{on}}), &
T_{\mathrm{on}}  \le t \le T
\end{cases}
$$

Para obtener el valor medio $I_{DC} = \left\langle i(t) \right\rangle$, mediante análisis geométrico, se debe calcular él área, que puede descomponerse en:

* Un rectángulo de altura $i_{\min}$ y base $T$, $\qquad A_{\text{rect}} = T \, i_{\min}$.
* Un triángulo de base $T$ y altura $\qquad \Delta i = i_{\max} - i_{\min}$, $A_{\text{tri}} = \frac{1}{2} \, T \, \Delta i$.

> Área total de la señal, $A_{\text{total}}=A_{\text{rect}}+A_{\text{tri}}=T\, i_{\min}+\frac{1}{2} T \Delta i = T\left(i_{\min}+\frac{\Delta i}{2}\right)$. El valor promedio es el área dividida entre el período:

$$
I_{DC}
=
\frac{A_{\text{total}}}{T}
=
i_{\min}
+
\frac{\Delta i}{2}
$$

Como $\Delta i = i_{\max} - i_{\min}$, también puede escribirse como:

$$
\boxed{
I_{DC}
=
\frac{i_{\max} + i_{\min}}{2}
}
$$

Este método es completamente geométrico y coincide con el cálculo integral del valor promedio.





```python
# Parámetros de la señal triangular
i_min         = 6         # corriente mínima [A]
i_max         = 10.0        # corriente máxima [A]
Delta_i       = i_max - i_min     # ripple [A]

D             = 0.25        #Duty cycle

Ton           = D*T        # tiempo de subida
Toff          = T-Ton      # tiempo de bajada
```

```python
# Construcción de la señal triangular
tau           = t % T # Operador módulo para repetir la señal

# Señal triangular periódica
i             = np.where(tau < Ton,  i_min + (Delta_i/Ton)*tau,  i_max - (Delta_i/Toff)*(tau - Ton))
```

```python
# Valor medio numérico
i_avg_num     = np.mean(i)

# Valor medio y RMS analítico
i_avg_ana     = i_min + Delta_i/2
```

```python
print(f"i_DC numérico  = {i_avg_num:.2f} A")
print(f"i_DC analítico = {i_avg_ana:.2f} A")
```

```python

# Gráfica de la señal
plt.plot(t*1e3, i, label="i(t) triangular")
plt.plot(t*1e3, i_avg_ana*np.ones_like(t),'r--', linewidth=3, label="Valor medio I_DC") # DC
plt.xlabel("Tiempo [ms]")
plt.ylabel("Corriente [A]")
plt.grid(True)
plt.legend()
plt.title("Corriente  periódica")
plt.show()
```

##  Ejemplo 5.3 — Valor medio de una corriente triangular activa durante una fracción $D$ del periodo

Considere una señal de corriente triangular, $i(t)$, periodica, $T$, durante la primera parte del tiempo, de duración $T_{\mathrm{on}}$, la corriente crece linealmente desde $i_{\min}$ hasta $i_{\max}$. Durante el resto del tiempo, de duración $T_{\mathrm{off}}$, la corriente cae a cero. La señal se repite cada periodo $T = T_{\mathrm{on}} + T_{\mathrm{off}}$

$$
i(t)=
\begin{cases}
i_{\min} + \frac{\Delta i}{T_{\mathrm{on}}}\, t \qquad & 0 \le t \le T_{\mathrm{on}}\\
0,\qquad & T_{\mathrm{on}} \le t \le T
\end{cases}
$$
Mediante análisis geométrico, obtener el valor medio $I_{DC} = \left\langle i(t) \right\rangle.$ Todo el cálculo es igual que en el ejercicio anterior, excepto que al calcular el área, la base es $T_{on}$ en vez de $T$.

* Un rectángulo de altura $i_{\min}$ y base $T_{on}$, $\qquad \qquad A_{\text{rect}} = T_{on} \, i_{\min}$.
* Un triángulo de base $T_{on}$ y altura $\quad \Delta i = i_{\max} - i_{\min}$, $\quad A_{\text{tri}} = \frac{1}{2} \, T_{on} \, \Delta i$.

> Área total de la señal, $A_{\text{total}}=A_{\text{rect}}+A_{\text{tri}}=T_{on}\, i_{\min}+\frac{1}{2} T_{on} \Delta i = T_{on}\left(i_{\min}+\frac{\Delta i}{2}\right)$. El valor promedio es el área dividida entre el período:

$$
\boxed{
I_{ON}
=
\frac{A_{\text{total}}}{T}
=
(i_{\min} + \frac{\Delta i}{2})·D
=
(\frac{i_{\max} + i_{\min}}{2})·D
=
I_{ON,DC}·D
}
$$

La corriente $I_{ON,DC}$ representa el valor medio de la señal calculado solo para el intervalo de tiempo $ON$. El valor medio se calcula respecto al período completo $T$, no solamente durante el intervalo en que la corriente es diferente de cero.




```python
# Parámetros de la señal triangular
i_min         = 6         # corriente mínima [A]
i_max         = 10.0        # corriente máxima [A]
Delta_i       = i_max - i_min     # ripple [A]

D             = 0.125        #Duty cycle

Ton           = D*T        # tiempo de subida
Toff          = T-Ton      # tiempo de bajada
```

```python
# Construcción de la señal triangular
tau           = t % T # Operador módulo para repetir la señal

# Señal triangular periódica
i             = np.where(tau < Ton,  i_min + (Delta_i/Ton)*tau,  0)
```

```python
# Valor medio numérico
i_avg_num     = np.mean(i)

# Valor medio analítico
i_avg_ana     = (i_min + Delta_i/2)*D
```

```python
print(f"i_DC numérico  = {i_avg_num:.2f} A")
print(f"i_DC analítico = {i_avg_ana:.2f} A")
```

```python

# Gráfica de la señal
plt.plot(t*1e3, i, label="i(t) triangular")
plt.plot(t*1e3, i_avg_ana*np.ones_like(t),'r--', linewidth=3, label="Valor medio I_DC") # DC
plt.xlabel("Tiempo [s]")
plt.ylabel("Corriente [A]")
plt.grid(True)
plt.legend()
plt.title("Corriente  periódica")
plt.show()
```

## Ejemplo 6. Señal de tres niveles de amplitud

A partir de la combinación de dos señales cuadradas bipolares de tensión se construye una primera aproximación de una señal escalonada de tres niveles que será de utilidad más adelante.

```python
# Parámetros de la señal
Vp          = 325       # V
f           = 50        # Hz
T           = 1 / f     # s
# Generación de la primera componente
v1          = np.where((t % T) < T/2, Vp, -Vp)
# Generación de la segunda componente con defase phi
k           = 0.375          # por ejemplo, 25% de desfase de periodo
phi         = k * T         # desfase en segundos

v2          = np.where(((t + phi) % T) < T/2, Vp, -Vp)
# Señal de tres niveles como resta de las dos cuadradas
v3          = v1 - v2
#  Normalizamos la señal
v3_norm     = v3 / 2
```

```python
# Tensión media
Vmean        = np.mean(v3)
```

```python
# Imprimir resultados
print(f"V_p = {Vp:.2f} V")
print(f"V_avg numérico = {Vmean:.4f} V")
```

```python
# Gráfica
plt.plot(t, v_seno, label="Señal senoidal")
plt.plot(t, v3_norm, label="Señal de tres niveles")
plt.title("Señal periódica: 0 → +Vp → 0 -Vp →  (DC = 0)")
plt.xlabel("Tiempo (s)")
plt.ylabel("v(t)")
plt.show()
```
