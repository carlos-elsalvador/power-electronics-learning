# **Cálculo del valor eficaz de una señal triangular de corriente**

![Figura 1. Señal Triangular alterna pura](figures/Fig1_ucm_rms.png)


La corriente es una señal triangular periódica que, durante un periodo, aumenta linealmente desde $-i_{AC}$ hasta $+i_{AC}$. Por tanto, $i_{AC}$ representa la amplitud de la señal (la mitad del rizado pico a pico):

$$\Delta i = 2i_{AC}$$

El periodo es $T$. Se resolverá el ejercicio en cinco pasos.

## PASO 1. Obtener la expresión de $i(t)$ para un periodo

Como la señal es una recta, escribimos: $i(t)=at+b$. Usamos los extremos de un periodo: $i(0)=-i_{AC}$ e $ i(T)=+i_{AC}$.

* Al sustituir $t=0$: $-i_{AC}=a\cdot 0+b \quad\Rightarrow\quad b=-i_{AC}$

* Al sustituir $t=T$: $i_{AC}=aT-i_{AC} \quad\Rightarrow\quad a=\frac{2i_{AC}}{T}$

* Por tanto, la expresión válida para $0\leq t\leq T$ es:$\qquad \boxed{i(t)=-i_{AC}+\frac{2i_{AC}}{T}t}$

## PASO 2. Calcular el área bajo $i(t)$ en un periodo

El área bajo la curva es la integral, corresponde a la integral de la corriente durante un periodo:

$$A=\int_0^T i(t)\,dt \quad =\int_0^T\left(-i_{AC}+\frac{2i_{AC}}{T}t\right)dt \quad =\left[-i_{AC}t+\frac{i_{AC}}{T}t^2\right]_0^T \quad =-i_{AC}T+i_{AC}T \quad=0$$


El área es cero porque la señal es simétrica respecto al eje horizontal: las áreas positiva y negativa se cancelan.

## PASO 3. Calcular el área bajo $i^2(t)$ en un periodo

* Elevamos al cuadrado la expresión de la corriente: $i^2(t)=\left(-i_{AC}+\frac{2i_{AC}}{T}t\right)^2$

* Desarrollando el cuadrado:$i^2(t)=i_{AC}^2\left(1-\frac{4t}{T}+\frac{4t^2}{T^2}\right)$

* El área bajo $i^2(t)$ durante un periodo es: $A=\int_0^T i^2(t)\,dt$

$$A=i_{AC}^2\int_0^T\left(1-\frac{4t}{T}+\frac{4t^2}{T^2}\right)dt \quad =i_{AC}^2\left[t-\frac{2t^2}{T}+\frac{4t^3}{3T^2}\right]_0^T \quad=i_{AC}^2\left(T-2T+\frac{4T}{3}\right) \quad=\frac{i_{AC}^2T}{3}$$

## PASO 4. Calcular el valor medio de $i^2(t)$

El valor medio de una función periódica durante un periodo es su integral dividida entre el periodo:

$$\left\langle i^2 \right\rangle=\frac{1}{T}\int_0^T i^2(t)\,dt =\quad \frac{1}{T}\cdot\frac{i_{AC}^2T}{3} \quad =\frac{i_{AC}^2}{3}$$

## PASO 5. Calcular el valor eficaz al cuadrado y el valor eficaz de $i(t)$

Por definición, el valor eficaz al cuadrado es el valor medio de la señal elevada al cuadrado: $I_{\mathrm{rms}}^2=\frac{1}{T}\int_0^T i^2(t)\,dt$. Este resultado es exactamente el obtenido en el paso 4. Por tanto:$I_{\mathrm{rms}}^2=\frac{i_{AC}^2}{3}$. Si además se desea el valor eficaz de la corriente, tomamos la raíz cuadrada positiva:

$$\boxed{I_{\mathrm{rms}}=\frac{i_{AC}}{\sqrt{3}}}$$

Como en la figura $i_{AC}=2.5\,\mathrm{A}$, se obtiene:$I_{\mathrm{rms}}^2=\frac{(2.5\,\mathrm{A})^2}{3}\approx2.083\,\mathrm{A}^2$ y, por tanto:$I_{\mathrm{rms}}=\frac{2.5}{\sqrt{3}}\,\mathrm{A}\approx1.443\,\mathrm{A}$

# **Cálculo RMS de una señal triangular de corriente con nivel DC**

![Figura 2. Señal Triangular alterna con componente DC](figures/Fig2_ucm_rms.png)

Sumamos un valor constante ($I_{DC}$, nivel DC) a la señal del ejemplo anterior. Seguir el mismo análisis y obtener valor RMS.

## PASO 1. Obtener la expresión de $i(t)$ para un periodo

Vamos a llamar $i_{LIN}(t)$ al resultado del paso 1 del ejemplo anterior, $i_{LIN}(t)=-i_{AC}+\frac{2i_{AC}}{T}t$. Ahora la expresión de $i(t)$ es exactamente la misma, pero sumando el nivel DC. Por lo tanto:
$$i(t)  =I_{DC}-i_{AC}+\frac{2i_{AC}}{T}t$$

## PASO 2. Calcular el área bajo $i^2(t)$ en un periodo

* Conocemos $i(t)= I_{DC} + i_{LIN}(t)$

* Conocemos $\int_0^Ti_{LIN}(t) = 0$

* Conocemos $\int_0^T i_{LIN}^2(t) = \frac{i_{AC}^2}{3}T$

La integral de la señal $i^2(t)$ queda como a continuación:
$$A=\int_0^T i(t) dt \quad = \int_0^T \left[ I_{DC} + I_{LIN}(t) \right]^2 \quad = \int_0^T \left[ I_{DC}^2 + 2I_{DC}I_{LIN}(t) + I_{LIN}^2(t) \right]dt$$


$$A= \int_0^T  I_{DC}^2 dt + \int_0^T 2I_{DC}I_{LIN}(t) dt + \int_0^T I_{LIN}^2(t)dt \quad = I_{DC}^2 T + 0 + \frac{i_{AC}^2}{3}T$$



## PASO 3. Calcular el valor medio de $i^2(t)$

El valor medio de una función periódica durante un periodo es su integral dividida entre el periodo, a la que llamaremos $I_{RMS}^2$:

$$I_{RMS}^2 = \left\langle i(t)^2 \right\rangle=\frac{1}{T}\int_0^T i^2(t)\,dt \quad =\frac{Área}{T}   = \frac{1}{T}\cdot \left( I_{DC}^2T +\frac{i_{AC}^2T}{3}\right) \quad =I_{DC}^2 +\frac{i_{AC}^2}{3}$$

Recordar que para una señal que combina un nivel DC y una señal AC pura, el valor eficaz al cuadrado es el nivel DC al cuadrado más el valor eficaz al cuadrado de la señal AC. Esto se evidencia en el desarrollo del paso anterior, en el que tenemos en cuenta que el promedio de la señal AC es nulo.

## PASO 4. Calcular el valor eficaz de $i(t)$

Se obtendrán una expresión general para $I_{RMS}$ en función de $I_{DC}$ e $i_{AC} = \frac{\Delta i}{2}$.
* Sabemos que $I_{DC} = \frac{(i_{MAX}+ i_{MIN})}{2}$
* Sabemos que $\Delta i = i_{MAX}- i_{MIN}$
* Sabemos que $i_{AC} = \frac{\Delta i}{2}$



$$I_{RMS}^2 = I_{DC}^2 +\frac{i_{AC}^2}{3} = I_{DC}^2 +\frac{(\frac{\Delta i}{2})^2}{3}  = I_{DC}^2 +\frac{(\Delta i)^2}{12} \qquad \boxed{I_{\mathrm{RMS}}= \sqrt{I_{DC}^2 + \frac{(\Delta i)^2}{12}\quad}} $$




#  **Cálculo RMS de señal de corriente triangular activa durante una fracción $D$ del periodo**

![Figura 3. Señal Triangular activa Ton](figures/Fig3_ucm_rms.png)

Considere una señal de corriente triangular, $i(t)$, periodica, $T$, durante la primera parte del tiempo, de duración $T_{\mathrm{on}}$, la corriente crece linealmente desde $i_{ON, \min}$ hasta $i_{ON, \max}$. Durante el resto del tiempo, de duración $T_{\mathrm{off}}$, la corriente cae a cero. La señal se repite cada periodo $T = T_{\mathrm{on}} + T_{\mathrm{off}}$




## PASO 1. Obtener la expresión de $i(t)$ para un periodo

$$
i(t)=
\begin{cases}
i_{ON, MIN} + \frac{\Delta i}{T_{\mathrm{on}}}\, t \qquad & 0 \le t \le T_{\mathrm{on}}\\
0,\qquad & T_{\mathrm{on}} \le t \le T
\end{cases}
$$

También,
* Sabemos que $I_{ON,DC} = \frac{(i_{ON,MAX}+ i_{ON,MIN})}{2}$
* Sabemos que $\Delta i = i_{ON,MAX}- i_{ON,MIN}$
* Sabemos que $i_{AC} = \frac{\Delta i}{2}$

Por ello, de manera equivalente, la expresión anterior se puede reescribir como:
$$
i(t)=
\begin{cases}
I_{ON, DC} - i_{AC}+\frac{2i_{AC}}{T_{on}}\, t \qquad & 0 \le t \le T_{\mathrm{on}}\\
0,\qquad & T_{\mathrm{on}} \le t \le T
\end{cases}
$$


# PASO 2 Calcular área y valor medio
Mediante análisis geométrico, obtener el valor medio $I_{DC} = \left\langle i(t) \right\rangle.$ Todo el cálculo es igual que  el realizado en cualquier ejercicio de cálculo del valor medio, excepto que al calcular el área, la base es $T_{on}$ en vez de $T$.

* Un rectángulo de altura $i_{ON, MIN}$ y base $T_{on}$, $\qquad \qquad A_{\text{rect}} = T_{on} \, i_{ON, MIN}$.
* Un triángulo de base $T_{on}$ y altura $ \Delta i $, $\qquad \qquad \qquad A_{\text{tri}} = \frac{1}{2} \, T_{on} \, \Delta i$.

> Área total de la señal, $A_{\text{total}}=A_{\text{rect}}+A_{\text{tri}}=T_{on}\, i_{ON, MIN}+\frac{1}{2} T_{on} \Delta i = T_{on}\left(i_{ON, MIN}+\frac{\Delta i}{2}\right)$. El valor promedio es el área dividida entre el período:

$$
I_{ON}
=
\frac{A_{\text{total}}}{T}
=
(i_{ON, MIN} + \frac{\Delta i}{2})·D
=
(\frac{i_{ON, MAX} + i_{ON, MIN}}{2})·D
=
\boxed{
I_{ON,DC}·D
}
$$

La corriente $I_{ON,DC}$ representa el valor medio de la señal calculado solo para el intervalo de tiempo $ON$. El valor medio se calcula respecto al período completo $T$, no solamente durante el intervalo en que la corriente es diferente de cero.

## PASO 3. Calcular el área bajo $i^2(t)$ en un periodo

Para calcular el área hay que integrar la señal $i^2(t)$ entre $0$ y $T$. Como a partir de $t = T_{ON}$ la corriente es nula, podemos hacer la integral entre $0$ y $T_{ON}$, ahora en vez de $T$ tendremos $T_{ON}$, así que la integral de la señal $i^2(t)$ queda como a continuación:

$$A=\int_0^{DT} i^2(t) dt \quad = \int_0^{DT} \left[ I_{ON,DC} + I_{LIN}(t) \right]^2 dt$$

Donde $i_{LIN}(t)$ es la componente de alterna, $ i_{LIN}(t)= -i_{AC}+\frac{2i_{AC}}{T_{on}}\, t $

$$A= \int_0^{DT} \left[ I_{ON,DC}^2 + 2I_{ON,DC}I_{LIN}(t) + I_{LIN}^2(t) \right]dt \\= \int_0^{DT}  I_{ON,DC}^2 dt + \int_0^{DT} 2I_{ON,DC}I_{LIN}(t) dt + \int_0^{DT} I_{LIN}^2(t)dt \\ = I_{ON,DC}^2 DT + 0 + \frac{i_{AC}^2}{3}DT$$




## PASO 4. Calcular el valor medio de $i^2(t)$

El valor medio de una función periódica durante un periodo es su integral dividida entre el periodo:


El valor medio de una función periódica durante un periodo es su integral dividida entre el periodo, a la que llamaremos $I_{RMS}^2$:

$$I_{RMS}^2 = \left\langle i(t)^2 \right\rangle=\frac{1}{T}\int_0^T i^2(t)\,dt \quad =\frac{Área}{T}   = \frac{1}{T}\cdot \left( I_{ON,DC}^2 DT +\frac{i_{AC}^2 DT}{3}\right) \quad =D·\left(I_{ON,DC}^2 +\frac{i_{AC}^2}{3}\right)$$

Para una señal triangular (o lineal) activa durante una fracción D del periodo, el valor eficaz al cuadrado se obtiene igual que si la señal ocupara el periodo completo (calculando los niveles DC y AC para el periodo en que la señal está activa) y se multiplica por D.
